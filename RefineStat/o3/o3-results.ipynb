```python
import pymc as pm
import numpy as np
import arviz as az
import matplotlib.pyplot as plt

# Given Data
N = 12  # Number of observations
n = np.array([47, 148, 119, 810, 211, 196, 148, 215, 207, 97, 256, 360])
r = np.array([0, 18, 8, 46, 8, 13, 9, 31, 14, 8, 29, 24])

with pm.Model() as m:
    # Hyperpriors for global log-odds and between-hospital variability
    alpha = pm.Normal("alpha", mu=0.0, sigma=5.0)
    sigma_h = pm.HalfNormal("sigma_h", sigma=2.0)

    # Hospital-specific random effects (standardized)
    theta_offset = pm.Normal("theta_offset", mu=0.0, sigma=1.0, shape=N)
    logit_p = alpha + sigma_h * theta_offset
    p = pm.Deterministic("p", pm.math.sigmoid(logit_p))

    # Likelihood: observed mortalities
    y_obs = pm.Binomial("y_obs", n=n, p=p, observed=r)

    # Sample from the posterior
    idata = pm.sample(
        1000,
        tune=1000,
        chains=4,
        return_inferencedata=True,
        idata_kwargs={"log_likelihood": True}
    )

    # Posterior diagnostics
    summary = az.summary(idata)
    loo_result = az.loo(idata)
    loo_elpd = loo_result.elpd_loo
    print(f"ELPD LOO: {loo_elpd}")
    ## Compute the ELPD LOO

    print(summary)
```

### GP

```python
import pymc as pm
import numpy as np
import arviz as az
import matplotlib.pyplot as plt

# Given Data
N = 11  # Number of observations
x = np.array([-10, -8, -6, -4, -2, 0, 2, 4, 6, 8, 10])
y = np.array([4.75906, 1.59423, 2.99548, 5.27501, 1.66472, 2.24347, 2.8914, 4.08681, 4.60588, 0.802364, 3.92136])
k = np.array([40, 37, 29, 12, 4, 3, 9, 19, 77, 82, 33])

with pm.Model() as m:
    # Data containers
    x_data = pm.Data("x_data", x[:, None])
    k_data = pm.Data("k_data", k)

    # GP hyperpriors
    ℓ = pm.Gamma("ℓ", alpha=2, beta=1)
    η = pm.HalfNormal("η", sigma=1)

    # Covariance function
    cov = η**2 * pm.gp.cov.ExpQuad(1, ls=ℓ)

    # Latent GP
    gp = pm.gp.Latent(cov_func=cov)
    f = gp.prior("f", X=x_data)

    # Link to Poisson rate
    mu = pm.Deterministic("mu", pm.math.exp(f))

    # Poisson likelihood
    y_obs = pm.Poisson("y_obs", mu=mu, observed=k_data)

    # Posterior sampling
    idata = pm.sample(
        1000,
        tune=1000,
        chains=4,
        return_inferencedata=True,
        idata_kwargs={"log_likelihood": True}
    )
    # Posterior diagnostics
    summary = az.summary(idata)
    loo_result = az.loo(idata)
    loo_elpd = loo_result.elpd_loo
    print(f"ELPD LOO: {loo_elpd}")
    ## Compute the ELPD LOO

    print(summary)
```

```python
import pymc as pm
import numpy as np
import arviz as az
import matplotlib.pyplot as plt

# Data
y = np.array([28, 8, -3, 7, -1, 1, 18, 12])
sigma = np.array([15, 10, 16, 11, 9, 11, 10, 18])

with pm.Model() as m:
    # Hyperpriors for the population mean and between‐school variability
    mu = pm.Normal("mu", mu=0, sigma=5)
    tau = pm.HalfCauchy("tau", beta=5)

    # School‐specific effects
    theta = pm.Normal("theta", mu=mu, sigma=tau, shape=y.shape[0])

    # Likelihood
    y_obs = pm.Normal("y_obs", mu=theta, sigma=sigma, observed=y)

    # Sample from the posterior
    idata = pm.sample(
        1000,
        tune=1000,
        chains=4,
        return_inferencedata=True,
        idata_kwargs={"log_likelihood": True}
    )

    # Posterior diagnostics
    summary = az.summary(idata)
    loo_result = az.loo(idata)
    loo_elpd = loo_result.elpd_loo
    print(f"ELPD LOO: {loo_elpd}")
    ## Compute the ELPD LOO

    print(summary)
```
